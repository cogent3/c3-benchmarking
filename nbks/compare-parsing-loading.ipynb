# Notation

- `bp` - biopython
- `c3` - cogent3
- `c3h5s` - sparse storage from cogent3-h5seqs plugin
- `sb` - scikit-bio

```python
import pathlib
from c3bench.report import tool_summary, data_summary, extract_function_snippets

RESULT_DIR = pathlib.Path("~/repos/gav_blog/docs/posts/benchmark_parsers").expanduser()

plot_settings = {
    "horizontal_spacing": 0.1,
    "title_font_size": 26,
    "axis_font_size": 18,
    "plot_title_font_size": 30,
}
layout_kwargs = {
    "width": 1000,
    "height": 500,
    "margin": {"l": 20, "r": 20, "t": 100, "b": 20},
}

summary = tool_summary("parse-fasta", "parse-gbk", "parse-ann-gff", "parse-ann-gb")
outpath = RESULT_DIR / "tools_summary.txt"
summary.write(outpath, format_name="md")
summary = data_summary("~/repos/c3-benchmarking/data", ["fa", "gb", "dat", "gff3"])
summary = summary.filtered(lambda x: x != "sars_msa", columns="dataset")
outpath = RESULT_DIR / "datasets_summary.txt"
summary.write(outpath, format_name="md")

# fasta format parsers
snipps = extract_function_snippets("c3bench.parse_fa")
outpath = RESULT_DIR / "parse_fa_snippets.txt"
outpath.write_text(snipps)

# fastq seq format parsers
snipps = extract_function_snippets("c3bench.parse_fq")
outpath = RESULT_DIR / "parse_fq_snippets.txt"
outpath.write_text(snipps)

# genbank seq format parsers
snipps = extract_function_snippets("c3bench.parse_gbk")
outpath = RESULT_DIR / "parse_gbk_snippets.txt"
outpath.write_text(snipps)

# parser gff3 annotations
snipps = extract_function_snippets("c3bench.parse_ann_gff")
outpath = RESULT_DIR / "parse_ann_gff_snippets.txt"
outpath.write_text(snipps)

# parser genbank annotations
snipps = extract_function_snippets("c3bench.parse_ann_gb")
outpath = RESULT_DIR / "parse_ann_gb_snippets.txt"
outpath.write_text(snipps)
```

# Comparing performance of raw parsers of basic data formats

## Parse Fasta

### Chimpanzee genome

```python
from c3bench.report import display_results_for

table, fig = display_results_for(
    "parse_fasta", "ptro_fa", title="<b>Parsing the Chimp genome</b>", **plot_settings
)
table.write(RESULT_DIR / "parse_fasta_ptro.txt", format_name="md")
table
```

```python
fig.layout.update(**layout_kwargs)
fig.show()
fig.write_image(RESULT_DIR / "parse_fasta_ptro.svg")
```
